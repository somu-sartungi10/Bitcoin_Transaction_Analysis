```python
from ml_engine.feature_extractor import ForensicFeatureExtractor
from ml_engine.model import ForensicAnomalyDetector
from ml_engine.evaluator import MLEvaluator
```

```python
extractor = ForensicFeatureExtractor()

x_test_features = extractor.fit_transform(x_test)
print(x_test_features.shape)
x_test_features.head()
```

```python
from pathlib import Path
from ml_engine.model import ForensicAnomalyDetector

model_dir = Path("../../model")

detector = ForensicAnomalyDetector()
detector.load(model_dir)
```

```python
# Recreate the calibration boundaries used during training
X_scaled = detector.scaler.transform(x_test_features)

raw_scores = detector.model.score_samples(X_scaled)

detector._min_raw_score = float(raw_scores.min())
detector._max_raw_score = float(raw_scores.max())

print("Min raw score:", detector._min_raw_score)
print("Max raw score:", detector._max_raw_score)
```

```python
risk_scores = detector.predict_risk_scores(x_test_features)

print(risk_scores[:50])
```

```python
print("Minimum:", risk_scores.min())
print("Maximum:", risk_scores.max())
print("Mean:", risk_scores.mean())
```

```python
y_true = y_test.to_numpy()

print("Actual anomalies:", y_true.sum())
print("Actual normal:", (y_true == 0).sum())
```

```python
threshold = 50.0

metrics = MLEvaluator.evaluate(
    y_true=y_true,
    risk_scores=risk_scores,
    threshold=threshold
)
```

```python
import pandas as pd

results = pd.DataFrame({
    "actual": y_true,
    "risk_score": risk_scores
})

print("ANOMALY SCORE DISTRIBUTION")
print(
    results[results["actual"] == 1]["risk_score"]
    .describe()
)

print("\nNORMAL SCORE DISTRIBUTION")
print(
    results[results["actual"] == 0]["risk_score"]
    .describe()
)
```

```python
for threshold in [50, 53, 54, 54.5, 55, 56, 60, 70, 75, 80, 90]:

    y_pred = (risk_scores >= threshold).astype(int)

    tp = ((y_true == 1) & (y_pred == 1)).sum()
    fp = ((y_true == 0) & (y_pred == 1)).sum()
    fn = ((y_true == 1) & (y_pred == 0)).sum()
    tn = ((y_true == 0) & (y_pred == 0)).sum()

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = (
        2 * precision * recall / (precision + recall)
        if (precision + recall) > 0
        else 0
    )

    print(
        f"Threshold={threshold:>5} | "
        f"Precision={precision:.4f} | "
        f"Recall={recall:.4f} | "
        f"F1={f1:.4f} | "
        f"TP={tp} FP={fp} FN={fn} TN={tn}"
    )
```
